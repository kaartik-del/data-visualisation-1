```python
import csv

# WRITE to a CSV file
data = [
    ['Name', 'Age', 'City'],
    ['Alice', '25', 'New York'],
    ['Bob', '30', 'London']
]

with open('people.csv', 'w', newline='') as file:
    writer = csv.writer(file)
    writer.writerows(data)

# READ from a CSV file
with open('people.csv', 'r') as file:
    reader = csv.reader(file)
    for row in reader:
        print(row)
```

```python
import json

data = [
    {"Name": "Alice", "Age": 25, "City": "New York"},
    {"Name": "Bob", "Age": 30, "City": "London"}
]

# WRITE to JSON file
with open('people.json', 'w') as file:
    json.dump(data, file, indent=4)

# READ from JSON file
with open('people.json', 'r') as file:
    loaded_data = json.load(file)
    for row in loaded_data:
        print(row)
```

```python
import xml.etree.ElementTree as ET

# WRITE to XML file
root = ET.Element("people")

p1 = ET.SubElement(root, "person")
ET.SubElement(p1, "Name").text = "Alice"
ET.SubElement(p1, "Age").text = "25"
ET.SubElement(p1, "City").text = "New York"

p2 = ET.SubElement(root, "person")
ET.SubElement(p2, "Name").text = "Bob"
ET.SubElement(p2, "Age").text = "30"
ET.SubElement(p2, "City").text = "London"

tree = ET.ElementTree(root)
tree.write('people.xml', encoding='utf-8', xml_declaration=True)

# READ from XML file
tree = ET.parse('people.xml')
root = tree.getroot()

for person in root.findall('person'):
    name = person.find('Name').text
    age = person.find('Age').text
    city = person.find('City').text
    print([name, age, city])
```

```python
import openpyxl

data = [
    ['Name', 'Age', 'City'],
    ['Alice', 25, 'New York'],
    ['Bob', 30, 'London']
]

# WRITE to Excel (.xlsx)
wb = openpyxl.Workbook()
ws = wb.active
ws.title = "Sheet1"

for row in data:
    ws.append(row)

wb.save('people.xlsx')
print("Excel file created.")

# READ from Excel (.xlsx)
wb = openpyxl.load_workbook('people.xlsx')
ws = wb["Sheet1"]

for row in ws.iter_rows(values_only=True):
    print(list(row))
```

```python
import csv
import pandas as pd

# =====================================================================
# 1. INITIAL DATA
# =====================================================================
data_dict = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva', 'Frank'],
    'Age': [25, 30, 35, 28, 22, 40],
    'City': ['New York', 'London', 'Paris', 'Tokyo', 'Berlin', 'Sydney'],
    'Score': [85, 92, 78, 88, 95, 70],
    'Passed': [True, True, True, True, True, False]
}

data_list = [
    ['Name', 'Age', 'City', 'Score', 'Passed'],
    ['Alice', 25, 'New York', 85, True],
    ['Bob', 30, 'London', 92, True],
    ['Charlie', 35, 'Paris', 78, True],
    ['David', 28, 'Tokyo', 88, True],
    ['Eva', 22, 'Berlin', 95, True],
    ['Frank', 40, 'Sydney', 70, False]
]

# =====================================================================
# 2. CSV OPERATIONS USING PANDAS
# =====================================================================
print("=================== PANDAS CSV OPERATIONS ===================")

# Create DataFrame
df_orig = pd.DataFrame(data_dict)

# --- WRITE CSV FILES WITH DIFFERENT DELIMITERS ---
df_orig.to_csv('data_comma.csv', sep=',', index=False)       # Standard Comma
df_orig.to_csv('data_tab.tsv', sep='\t', index=False)         # Tab-Separated
df_orig.to_csv('data_pipe.txt', sep='|', index=False)         # Pipe-Separated
df_orig.to_csv('data_semicolon.csv', sep=';', index=False)   # Semicolon-Separated

# --- READ CSV FILES ---
df = pd.read_csv('data_comma.csv', sep=',')
df_tab = pd.read_csv('data_tab.tsv', sep='\t')
df_pipe = pd.read_csv('data_pipe.txt', sep='|')
df_semi = pd.read_csv('data_semicolon.csv', sep=';')

# --- DATAFRAME INSPECTION FUNCTIONS ---
print("1. Shape (Rows, Cols):", df.shape)
print("\n2. First 5 Rows (df.head()):\n", df.head(5))
print("\n3. Last 5 Rows (df.tail()):\n", df.tail(5))
print("\n4. Column Names:", df.columns.tolist())
print("\n5. First 5 Columns (df.iloc[:, :5]):\n", df.iloc[:, :5])
print("\n6. Select 'Name' Column:\n", df['Name'])
print("\n7. Summary Statistics:\n", df.describe())
print("\n8. DataFrame Info:")
df.info()


# =====================================================================
# 3. CSV OPERATIONS USING PURE PYTHON (NO PANDAS)
# =====================================================================
print("\n=================== PURE PYTHON CSV OPERATIONS ===================")

# --- WRITE CSV FILES WITH DIFFERENT DELIMITERS ---
with open('py_comma.csv', 'w', newline='') as f:
    csv.writer(f, delimiter=',').writerows(data_list)

with open('py_tab.tsv', 'w', newline='') as f:
    csv.writer(f, delimiter='\t').writerows(data_list)

with open('py_pipe.txt', 'w', newline='') as f:
    csv.writer(f, delimiter='|').writerows(data_list)

with open('py_semicolon.csv', 'w', newline='') as f:
    csv.writer(f, delimiter=';').writerows(data_list)

# --- READ CSV FILE ---
with open('py_comma.csv', 'r') as f:
    csv_data = list(csv.reader(f, delimiter=','))

headers = csv_data[0]
rows = csv_data[1:]

# --- PURE PYTHON INSPECTION FUNCTIONS ---
def get_shape(data):
    return (len(data) - 1, len(data[0]))

def head(data, n=5):
    return data[:n + 1]

def tail(data, n=5):
    return [data[0]] + data[-n:]

def select_column(data, col_name):
    idx = data[0].index(col_name)
    return [row[idx] for row in data[1:]]

def first_n_columns(data, n=5):
    return [row[:n] for row in data]

print("1. Shape:", get_shape(csv_data))
print("2. First 5 Rows (head):", head(csv_data, 5))
print("3. Last 5 Rows (tail):", tail(csv_data, 5))
print("4. Column Names:", headers)
print("5. First 5 Columns:", first_n_columns(csv_data, 5))
print("6. 'Name' Column:", select_column(csv_data, 'Name'))
```
